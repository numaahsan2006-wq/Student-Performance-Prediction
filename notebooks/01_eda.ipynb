```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Project root
PROJECT_ROOT = Path.cwd()

# Dataset
DATA_PATH = PROJECT_ROOT / "data" / "StudentPerformanceFactors.csv"

df = pd.read_csv("C:/Users/numaa/Downloads/StudentPerformanceFactors.csv")

print(f"Dataset shape: {df.shape}")

df.head()
```

```python
df.info()
```

```python
df.describe().T
```

```python
missing_values = df.isnull().sum()

missing_values = (
    missing_values[missing_values > 0]
    .sort_values(ascending=False)
)

missing_values
```

```python
plt.figure(figsize=(10, 5))

missing_values.plot(kind="bar")

plt.title("Missing Values by Feature")
plt.xlabel("Feature")
plt.ylabel("Number of Missing Values")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Exam_Score",
    bins=30,
    kde=True
)

plt.title("Distribution of Exam Scores")
plt.xlabel("Exam Score")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Attendance",
    y="Exam_Score",
    alpha=0.5
)

sns.regplot(
    data=df,
    x="Attendance",
    y="Exam_Score",
    scatter=False
)

plt.title("Attendance vs Exam Score")
plt.xlabel("Attendance (%)")
plt.ylabel("Exam Score")

plt.tight_layout()
plt.show()
```

```python
numerical_columns = df.select_dtypes(
    include=["int64", "float64"]
).columns

correlation_matrix = df[numerical_columns].corr()

plt.figure(figsize=(10, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Matrix")

plt.tight_layout()
plt.show()
```

```python
# Check for invalid exam scores
invalid_scores = df[
    (df["Exam_Score"] < 0) | (df["Exam_Score"] > 100)
]

print(f"Invalid exam score records: {len(invalid_scores)}")
print(invalid_scores[["Exam_Score"]])
```

```python
# Correct invalid exam scores to the valid 0–100 range
df["Exam_Score"] = df["Exam_Score"].clip(lower=0, upper=100)

print(
    "Maximum Exam Score after correction:",
    df["Exam_Score"].max()
)
```
